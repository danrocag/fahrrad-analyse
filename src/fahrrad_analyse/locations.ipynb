```python
import pandas as pd
import numpy as np
import seaborn as sns
import folium

df = pd.read_csv("../../data/location_data.csv", parse_dates=True)
df.head()
```

```python
df.latitude.min()
```

```python
m = folium.Map()
for _, row in df.iterrows():
    folium.CircleMarker([row.latitude, row.longitude], radius=4, popup=row.counter_site).add_to(m)

sw = [df.latitude.min(), df.longitude.min()]
ne = [df.latitude.max(), df.longitude.max()]
m.fit_bounds([sw, ne])

m
```
